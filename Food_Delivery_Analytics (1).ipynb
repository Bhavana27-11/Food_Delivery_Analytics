```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
customer_counts=df['Customer_Type'].value_counts()
print(customer_counts)
```

```python
#Customer type distribution
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='Customer_Type',order=df['Customer_Type'].value_counts().index)
plt.title('Order Distribution by Customer Type')
plt.xlabel('Customer Type')
plt.ylabel('Number of Orders')
plt.show()
```

```python
customer_revenue = df.groupby('Customer_Type')['Revenue_USD'].sum().sort_values(ascending=False)
print(customer_revenue)
```

```python
#Customer Type vs Revenue
```

```python
plt.figure(figsize=(8,5))
customer_revenue.plot(kind='bar')
plt.title('Revenue by Customer Type')
plt.xlabel('Customer Type')
plt.ylabel('Total Revenue (USD)')
plt.xticks(rotation=0)
plt.show()
```

```python
aov_customer = df.groupby('Customer_Type')['Order_Value_USD'].mean().sort_values(ascending=False)
print(aov_customer)
```

```python
#Average Order Value by Customer Type
```

```python
plt.figure(figsize=(8,5))
aov_customer.plot(kind='bar')
plt.title('Average Order Value by Customer Type')
plt.xlabel('Customer Type')
plt.ylabel('Average Order Value (USD)')
plt.xticks(rotation=0)
plt.show()
```

```python
rating_customer = df.groupby('Customer_Type')['Customer_Rating'].mean().sort_values(ascending=False)
print(rating_customer)
```

```python
#Customer Rating by Customer Type
```

```python
plt.figure(figsize=(8,5))
rating_customer.plot(kind='bar')
plt.title('Average Customer Rating by Customer Type')
plt.xlabel('Customer Type')
plt.ylabel('Average Rating')
plt.xticks(rotation=0)
plt.ylim(0,5)
plt.show()
```

```python
churn_customer = df.groupby('Customer_Type')['Churn_Risk'].mean().sort_values(ascending=False)
print(churn_customer)
```

```python
#Customer Type and Churn Risk
```

```python
plt.figure(figsize=(8,5))
churn_customer.plot(kind='bar')
plt.title('Average Churn Risk by Customer Type')
plt.xlabel('Customer Type')
plt.ylabel('Average Churn Risk')
plt.xticks(rotation=0)
plt.show()
```

```python
#Order Value Analysis
```

```python
plt.figure(figsize=(9,5))
sns.histplot(data=df,x='Order_Value_USD',bins=20,kde=True)
plt.title('Distribution of Order Value')
plt.xlabel('Order Value (USD)')
plt.ylabel('Number of Orders')
plt.show()
```

```python
#Delivery Performance
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df,x='Delivery_Distance_KM',y='Delivery_Time_Min',hue='Traffic_Level',alpha=0.6)
plt.title('Delivery Distance vs Delivery Time')
plt.xlabel('Delivery Distance (KM)')
plt.ylabel('Delivery Time (Minutes)')
plt.show()
```

```python
#Customer Experience
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='Traffic_Level',y='Customer_Rating')
plt.title('Customer Rating by Traffic Level')
plt.xlabel('Traffic Level')
plt.ylabel('Customer Rating')
plt.ylim(0, 5)
plt.show()
```

```python
#Restaurant & Cuisine Analysis
```

```python
cuisine_counts = df['Cuisine_Type'].value_counts()
plt.figure(figsize=(7,7))
plt.pie(cuisine_counts,labels=cuisine_counts.index,autopct='%1.1f%%',startangle=90)
plt.title('Order Share by Cuisine Type')
plt.show()
```

```python
#Profitability
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df,  x='Order_Value_USD',y='Profit_USD',hue='Customer_Type',alpha=0.25, s=20)

sns.regplot( data=df,x='Order_Value_USD',y='Profit_USD',scatter=False,color='black',line_kws={'linewidth': 2})
plt.axhline(0, linestyle='--', linewidth=1)
plt.title('Revenue vs Profit by Customer Type')
plt.xlabel('Order Value (USD)')
plt.ylabel('Profit (USD)')
plt.legend(title='Customer Type')
plt.tight_layout()
plt.show()
```

```python
#Monthly Trend 
```

```python
monthly_revenue = df.groupby('Month')['Revenue_USD'].sum()
plt.figure(figsize=(10,5))
monthly_revenue.plot(marker='o')
plt.title('Monthly Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Revenue (USD)')
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
#Customer Experience
```

```python
plt.figure(figsize=(8,5))
sns.violinplot( data=df,x='Customer_Type',y='Customer_Rating')
plt.title('Customer Rating Distribution by Customer Type')
plt.xlabel('Customer Type')
plt.ylabel('Customer Rating')
plt.show()
```

```python
# Count complaints and refunds
complaints = df['Complaint_Flag'].sum()
refunds = df['Refund_Flag'].sum()
plt.bar(['Complaints', 'Refunds'], [complaints, refunds])
plt.title('Complaints vs Refunds')
plt.ylabel('Number of Orders')
plt.show()

```
